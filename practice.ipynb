```python
num = int(input("Enter prime range:"))
flag = bool()
for i in range(num + 1):
    if i == 2:
        print(i)
    for j in range (2, i):
        if i%j == 0:
            flag = False
            break
        else:
            flag = True
    if flag:
        print(i)
    
```
